```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv("data.csv")
df
```

```python
%matplotlib inline
import matplotlib.pyplot as plt

plt.scatter(df.Hours_Studied, df.Marks_received,color='purple',marker='+')
plt.xlabel("Hours Studied")
plt.ylabel("Marks Received")
plt.title("Hours Studied vs Marks Received")
plt.show()
```

```python
from sklearn import linear_model

reg = linear_model.LinearRegression()
reg.fit(df[['Hours_Studied']], df['Marks_received'])
```

```python
reg.predict([[3.12]])
```

```python
%matplotlib inline
import matplotlib.pyplot as plt

plt.scatter(df.Hours_Studied, df.Marks_received,color='purple',marker='+')
plt.xlabel("Hours Studied")
plt.ylabel("Marks Received")
plt.title("Hours Studied vs Marks Received")
plt.show()
plt.plot(df.Hours_Studied, reg.predict(df[['Hours_Studied']],color='blue')
```
